## For in - Estrutura de repetição

```typescript
const frutas = ['Pêra', 'Maçã', 'Banana', 'Uva'];

for(let i = 0; i < frutas.length; i++){
    console.log(frutas[i]);
}
```

```typescript
const frutas = ['Pêra', 'Maçã', 'Banana', 'Uva'];

for(let i in frutas){
    console.log(i, frutas[i]);
}
```

```typescript
const pessoa = {
    nome:'Kauã',
    idade: 20,
    altura:1.75,
    endereço: {
        rua:'Rua Legal',
        numero: 123,
    },
};

console.log(pessoa['nome'])
console.log(pessoa.nome)

for (let i in pessoa){
    if (i === 'endereço'){
        for (let j in pessoa[i]){
            console.log(pessoa[i][j]);
        }
        break;
    }
    console.log(pessoa[i]);

}
```
