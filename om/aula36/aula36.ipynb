# DOM e árvore do DOM
Document Object Model -> É uma maneira de representar (uma interface/api) os objetos que compoem a estrutura conteúdo de uma documento Web. Ele é construido em cima de nós e objetos para que linguagens de programação possam alterar a estrutura, estilo e conteúdo de um documento HTML (ou XML). O DOM é a representação orientada a objetos da página web, que pode ser modificada por uma linguagem de scripts, como Javascript.

Existe um padrão específico para a implementação de um DOM nos navegadores, porém a maioria deles extendem o padrão (adicionam coisas a+).


```typescript
var paragraphs = document.getElementsByTagName("p");
// paragraphs[0] is the first <p> element
// paragraphs[1] is the second <p> element, etc.
alert(paragraphs[0].nodeName);
```

Como a grande maioria do código que usa o DOM gira em torno da manipulação de documentos HTML, é comum sempre se referir aos nós no DOM como elementos, pois em um documento HTML, cada nó é um elemento. Apesar de não ser estritamente precisa, a documentação que você encontrará no MDN frequentemente fará a mesma coisa, por causa de quão comum é essa suposição.

API -> é um conjunto de regras, protocolos e ferramentas que permite que diferentes componentes ou sistemas de software se comuniquem e interajam entre si

Node -> Node é uma interface da qual diversos tipos do DOM herdam, e que permite que esses tipos sejam tratados de forma similar, por exemplo, herdando os mesmos métodos ou sendo testados da mesma forma.
