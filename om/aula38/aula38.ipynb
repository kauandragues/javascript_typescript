## For of  e For each- Estrutura de Repetição

```typescript
const nome = 'Kauã Andrade';

for (let i of nome){
    console.log(i);
}
```

```typescript
const frutas = ['Pêra', 'Maçã', 'Banana', 'Uva'];

for (let i of frutas){
    console.log(i);
}
```

```typescript
const frutas = ['Pêra', 'Maçã', 'Banana', 'Uva'];

const array = frutas.forEach((valor, indice, array) => {
  console.log(valor, indice, array);
});
```
