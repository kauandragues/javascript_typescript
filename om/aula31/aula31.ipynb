## Diferenças entre `var` e `let/const`

### Semelhanças:

- Ambas criam variáveis;

```typescript
let nome = 'Luiz';
var nome2 = 'Luiz';
```

### Diferenças

- É possível redeclara apenas a variável `var`;
- A variável `var` fica disponível apenas na função ou global (se for declarada lá);
- A variável `let` fica disponível apenas dentro do escopo do BLOCO em que foi declarada;
- A variável `var` passa pelo processo de hoisting:

"Como as declarações de variáveis (e declarações em geral) são processadas antes de qualquer código seja executado, declarar uma variável em qualquer lugar no código é equivalente a declarar no inicio. Isso também significa que uma variável pode aparecer para ser usada antes dela ser declarada. Esse comportamento é chamado de "hoisting", a variável é movida para o inicio da função ou do código global."

```typescript
var nome2 = 'Kauã';
var nome2 = 'Pedro';
console.log(nome2);
```

```typescript
let nome = 'Kauã'; // let de fora do escopo do if
var nome2 = 'Andrade'; // var no escopo do módulo

if (true){

    // não é possível redeclarar variáveis let,
    // porém é como se fosse declarada um variável let apenas para esse escopo de bloco
    let nome = 'Pedro';
    if (true){
        let nome = 'Miguel';
        console.log(nome);
    }
    console.log(nome);

}
console.log(nome); 
// o let nome do escopo global é independente da variável nome declarada dentro do bloco if
//se eu quisesse acessar a variável nome de fora do escopo dentro do escopo, basta não redeclarar a variável
```

```typescript
var nome2 = 'Andrade'; // var no escopo global

if (true){

    // é possível redeclarar variáveis var, e ao fazer isso
    // a variável anterior é substituida pela atual
    // idependente do escopo em que ambas se encontram
    var nome2 = 'Pedro';
    if (true){
        var nome2 = 'Miguel';
        console.log(nome2);
    }
    console.log(nome2);

}
console.log(nome2);
```

```typescript
function mostraOi(){
    var mensagem = 'Oi';
    return mensagem;
}

mostraOi()
console.log(mensagem); // não tem acesso a var no escopo da função
```
