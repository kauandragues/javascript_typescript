## Atribuição via desestruturação (Objetos)

```typescript
const pessoa = {
    // nome: 'Luiz',
    sobrenome: 'Miranda',
    idade: 30,
    endereco: {
        rua: 'Rua qualquer',
        bairro: 'Bairro Qualquer',
        numero: '123',
    },
    restante: 323141,
};

const {nome = 'se não existe, use essa valor\n', sobrenome:outroNomeParaVariavel, idade, endereco:{rua, bairro, numero},...resto} = pessoa;
console.log(nome, outroNomeParaVariavel, idade, rua, bairro, numero);
console.log(resto);
```
